```python
# ==========================================
#   SISTEMA IA DE DENTISTAS
# ==========================================

class Dentista:

    def __init__(self):
        self.dentistas = []

    # FUNCION PARA AGREGAR DENTISTA
    def agregar_dentista(self):

        print("\n===== AGREGAR DENTISTA =====")

        colegiado = input("Numero de colegiado: ")
        clinica = input("Clinica propia: ")
        servicios = input("Servicios ofrecidos: ")
        materiales = input("Materiales en stock: ")
        citas = input("Citas del dia: ")
        anios = input("Anios de practica: ")

        datos = {
            "Colegiado": colegiado,
            "Clinica": clinica,
            "Servicios": servicios,
            "Materiales": materiales,
            "Citas": citas,
            "Anios": anios
        }

        self.dentistas.append(datos)

        print("\n✅ Dentista agregado correctamente.")

    # FUNCION PARA LISTAR DENTISTAS
    def listar_dentistas(self):

        print("\n=========== LISTA DE DENTISTAS ===========")

        if len(self.dentistas) == 0:
            print("No hay dentistas registrados.")

        else:

            print("-" * 120)

            print(f"{'Colegiado':<15}"
                  f"{'Clinica':<15}"
                  f"{'Servicios':<25}"
                  f"{'Materiales':<25}"
                  f"{'Citas':<10}"
                  f"{'Anios':<10}")

            print("-" * 120)

            i = 0

            while i < len(self.dentistas):

                print(f"{self.dentistas[i]['Colegiado']:<15}"
                      f"{self.dentistas[i]['Clinica']:<15}"
                      f"{self.dentistas[i]['Servicios']:<25}"
                      f"{self.dentistas[i]['Materiales']:<25}"
                      f"{self.dentistas[i]['Citas']:<10}"
                      f"{self.dentistas[i]['Anios']:<10}")

                i += 1

            print("-" * 120)

    # FUNCION PARA ACTUALIZAR DENTISTA
    def actualizar_dentista(self):

        print("\n===== ACTUALIZAR DENTISTA =====")

        colegiado = input("Ingresa el numero de colegiado a actualizar: ")

        encontrado = False
        i = 0

        while i < len(self.dentistas):

            if self.dentistas[i]["Colegiado"] == colegiado:

                print("\nNuevo dato del dentista")

                self.dentistas[i]["Clinica"] = input("Nueva clinica propia: ")
                self.dentistas[i]["Servicios"] = input("Nuevos servicios: ")
                self.dentistas[i]["Materiales"] = input("Nuevos materiales: ")
                self.dentistas[i]["Citas"] = input("Nuevas citas del dia: ")
                self.dentistas[i]["Anios"] = input("Nuevos anios de practica: ")

                print("\n✅ Dentista actualizado correctamente.")

                encontrado = True

            i += 1

        if encontrado == False:
            print("❌ No se encontro el dentista.")

    # FUNCION PARA BORRAR DENTISTA
    def borrar_dentista(self):

        print("\n===== BORRAR DENTISTA =====")

        colegiado = input("Ingresa el numero de colegiado a borrar: ")

        encontrado = False
        i = 0

        while i < len(self.dentistas):

            if self.dentistas[i]["Colegiado"] == colegiado:

                self.dentistas.pop(i)

                print("\n✅ Dentista eliminado correctamente.")

                encontrado = True

            else:
                i += 1

        if encontrado == False:
            print("❌ No se encontro el dentista.")


# ==========================
# PROGRAMA PRINCIPAL
# ==========================

sistema = Dentista()

opcion = 0

while opcion != 5:

    print("\n===================================")
    print("      SISTEMA IA DE DENTISTAS")
    print("===================================")
    print("1. Agregar dentista")
    print("2. Listar dentistas")
    print("3. Actualizar dentista")
    print("4. Borrar dentista")
    print("5. Salir")
    print("===================================")

    opcion = int(input("Selecciona una opcion: "))

    if opcion == 1:
        sistema.agregar_dentista()

    elif opcion == 2:
        sistema.listar_dentistas()

    elif opcion == 3:
        sistema.actualizar_dentista()

    elif opcion == 4:
        sistema.borrar_dentista()

    elif opcion == 5:
        print("\nGracias por usar el sistema IA de Dentistas 🦷")

    else:
        print("\n❌ Opcion no valida.")
```
